# Exploratory Data Analysis (EDA)

```python
raw_train.columns.values
```

```python
columns = ['V1', 'V2', 'V3', 'V4', 'V5', 'V6', 'V7', 'V8', 'V9', 'V10', 'V11', 'V12', 'V13', 'V14', 
           'V15', 'V16', 'V17', 'V18', 'V19', 'V20', 'V21', 'V22', 'V23', 'V24', 'V25', 'V26', 'V27',
           'V28', 'Amount']

ncols = 3
nrows = len(columns)

fig, axes = plt.subplots(nrows = nrows, ncols = ncols, figsize = (18, 5*nrows))

for r in range(nrows):
    column = columns[r]
    
    sns.histplot(raw_train[column], color = '#ffdbdb', label = 'Train', kde = False, stat = 'density', alpha = 0.5, ax = axes[r, 0])
    sns.histplot(raw_test[column], color = '#f6e0b5', label = 'Test', kde = False, stat = 'density', alpha = 0.5, ax = axes[r, 0])
    axes[r, 0].set_title(f'The distribution of {column}')
    axes[r, 0].legend()
    
    sns.boxplot(y = raw_train[column], color = '#ddc6c6', ax = axes[r, 1])
    axes[r, 1].set_title(f'Train Boxplot of {column}')
    
    sns.boxplot(y = raw_test[column], color = '#9799ba', ax = axes[r, 2])
    axes[r, 2].set_title(f'Test Boxplot of {column}')

    
    
plt.tight_layout()
plt.show()
```

```python
sns.countplot(data = raw_train, x = raw_train['Class'].astype('object'), palette = ['#d5d3dd', '#8d988d'])
plt.title('Number of Occurences for Class')

total = len(raw_train['Class'])

not_fraud_count = (raw_train['Class'] == 0).sum()
fraud_count = (raw_train['Class'] == 1).sum()

percentage_not_fraud = (not_fraud_count/total) * 100
percentage_fraud = (fraud_count/total) * 100

print(f'The percentage of Not Fraud (Class = 0): {percentage_not_fraud:.2f}%')
print(f'The percentage of Fraud (Class = 1): {percentage_fraud:.2f}%')

plt.tight_layout()
plt.show()
```

# Modelling

## Training Data

```python
model_train = raw_train.copy(deep = True)
model_test = raw_test.copy(deep = True)
```

```python
model_train.columns.values
```

```python
model_train = model_train.drop(['id', 'Time'], axis = 1)
model_train.info()
```

```python
X_train = model_train.drop(['Class'], axis = 1)
y_train = model_train['Class']
```

```python
# !pip install lightgbm
```

```python
# !pip install catboost
```

```python
from sklearn.model_selection import cross_val_score
import lightgbm as lgb
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
```

## LightGBM Model

```python
lgbm_model = lgb.LGBMClassifier(random_state = 42)

cv_lgbm = cross_val_score(lgbm_model, X_train, y_train, cv = 5, scoring = 'roc_auc')
```

```python
print("The ROC-AUC Cross-Validation Scores of LightGBM Model: ", cv_lgbm)
print("The Mean ROC-AUC of LightGBM Model: ", np.mean(cv_lgbm))
```

## XGBoost Model

```python
xgb_model = XGBClassifier(random_state = 42)

cv_xgb = cross_val_score(xgb_model, X_train, y_train, cv = 5, scoring = 'roc_auc')
```

```python
print("The ROC-AUC Cross-Validation Scores of XGBoost Model: ", cv_xgb)
print("The Mean ROC-AUC of XGBoost Model: ", np.mean(cv_xgb))
```

## CatBoost Model

```python
cat_model = CatBoostClassifier(verbose = 0, random_state = 42) # Suppress output for simplicity

cv_cat = cross_val_score(cat_model, X_train, y_train, cv = 5, scoring = 'roc_auc')
```

```python
print("The ROC-AUC Cross-Validation Scores of CatBoost Model: ", cv_cat)
print("The Mean ROC-AUC of CatBoost Model: ", np.mean(cv_cat))
```

## Test Data

### Best Model - CatBoost Model

```python
X_test = model_test.drop(['id', 'Time'], axis = 1)
X_test.columns.values
```

```python
cat_model.fit(X_train, y_train)
prediction_cat = cat_model.predict(X_test)
```

## Predicted Output

```python
test_predictions = pd.DataFrame({
    'id': model_test['id'],
    'Predicted Class': prediction_cat})

print(test_predictions.head())
```

## Feature Importances of CatBoost Model

```python
plt.figure(figsize = (10, 6))

feature_importances = cat_model.get_feature_importance()
sorted_feature = np.argsort(feature_importances)[::-1]

plt.barh(X_train.columns[sorted_feature], feature_importances[sorted_feature], color = '#e7d887')
plt.title('The Feature Importances Based On CatBoost Model')
plt.gca().invert_yaxis()
plt.show()
```
